## 1. Observability: is the loop fast enough?

Over time, we will use models, evaluate them, change models, and repeat. To evaluate and compare these models, we need to understand their performance. Token Factory has built-in observability to make this easy.

**The three latency numbers**:

- **TTFT:** *Time To First Token*. How 'snappy' and responsive the model feels. **When it matters:** a customer-facing chatbot, for example. If nothing appears for three seconds, the user assumes it's broken and leaves.
- **t/s:** *Tokens per Second*. How fast does the model feel while it is giving a response? **When it matters:** long generations streaming to a reader. A report being written on screen at 10 t/s feels slow and frustrating; at 60 t/s it feels alive.
- **E2EL:** *End-to-End Latency*. How long from request to finished output? **When it matters:** whenever nobody is watching the stream! In our agentic loops, we are not watching out model at all times. But it does not mean that the total response time does not matter - if our agent is slow, it won't be as effective.

## 2. Costs: is the loop cheap enough?

**LLMs can get expensive!** Every time the model is receiving input and producing output, it is actively billing. This is why cost management becomes **extremely important as our model scales**. Later, we will discuss best practices, so our agent does not break the bank.

The math itself is simple - the card gives you two prices, the response gives you two token counts:

$$\text{cost} = \frac{\text{prompt tokens} \times \text{price}_{in} + \text{completion tokens} \times \text{price}_{out}}{1{,}000{,}000}$$

A prototype request that costs a fraction of a cent feels free. The same request at 100,000 calls/day is a lot of money! When agents are running wild, we have less control over how much they use. Therefore, it is a great habit to **know what every request costs.** Let's price one real request:

````python
# Prices from the Nemotron 3 Nano model card, $/1M tokens
PRICE_IN = 0.06
PRICE_OUT = 0.24

resp = client.chat.completions.create(
    model="nvidia/NVIDIA-Nemotron-3-Nano-30B-A3B",
    messages=[{"role": "user", "content": "Summarize the outlook for the specialty coffee market in one paragraph."}],
)

u = resp.usage
cost = (u.prompt_tokens * PRICE_IN + u.completion_tokens * PRICE_OUT) / 1e6

print(f"input tokens:  {u.prompt_tokens}")
print(f"output tokens: {u.completion_tokens}")
print(f"cost:          ${cost:.6f}")
````

### Exercise: How is our bank feeling?

Calculate the cost per request for 2 models using the request above ( `messages=[{"role": "user", "content": "Summarize the outlook for the specialty coffee market in one paragraph."}]` ):

1. `nvidia/NVIDIA-Nemotron-3-Nano-30B-A3B` (prices above)
2. `nvidia/Nemotron-3-Ultra-550b-a55b` (you will need to find its pricing on the card!)

What is the average cost per request on each? Per 1,000 requests? Per 100,000? At what scale does the gap stop feeling academic?

````python
# Prices from each model card, $/1M tokens (input, output)
PRICES = {
    "nvidia/NVIDIA-Nemotron-3-Nano-30B-A3B": (0.06, 0.24),
    "nvidia/Nemotron-3-Ultra-550b-a55b": (1.00, 3.00),
}
PROMPT = "Summarize the outlook for the specialty coffee market in one paragraph."
RUNS = 5  # output length varies run to run, so average a few requests

avg_cost = {}
for model, (price_in, price_out) in PRICES.items():
    costs, out_tokens = [], []
    for _ in range(RUNS):
        r = client.chat.completions.create(model=model, messages=[{"role": "user", "content": PROMPT}])
        costs.append((r.usage.prompt_tokens * price_in + r.usage.completion_tokens * price_out) / 1e6)
        out_tokens.append(r.usage.completion_tokens)
    avg_cost[model] = sum(costs) / RUNS
    print(f"[bold]{model}[/bold]")
    print(f"  output tokens per run: {out_tokens}")
    print(f"  average cost/request:  ${avg_cost[model]:.6f}")

nano, ultra = avg_cost.values()
print(f"\nUltra costs {ultra / nano:.0f}× Nano per request on this prompt\n")
print(f"{'requests':>12} {'Nano':>12} {'Ultra':>12} {'gap':>12}")
for n in [1, 1_000, 100_000, 1_000_000]:
    print(f"{n:>12,} {'$' + format(nano * n, ',.2f'):>12} {'$' + format(ultra * n, ',.2f'):>12} {'$' + format((ultra - nano) * n, ',.2f'):>12}")
````

**Answer** (averages of 5 runs, from the cell above):

| | Nano | Ultra |
|---|---|---|
| Average output tokens | ~245 | ~335 |
| **Per request** | **\$0.000061** | **\$0.001033** (17×) |
| Per 1,000 requests | \$0.06 | \$1.03 |
| Per 100,000 requests | \$6.07 | \$103.32 |
| Per 1,000,000 requests | \$60.70 | \$1,033.20 |

**When does the gap stop feeling academic?** Around **100,000 requests a day**. The difference is about \$97 a day, roughly \$2,900 a month for the same summaries. At a million a day it's about \$29,000 a month. Agents get there sooner than you'd think: if one task takes 10 calls, 10,000 tasks a day already means 100,000 requests.

**The multiplier depends on the prompt.** On this short summary Ultra barely reasons, so the gap (17×) is close to the price gap. On Notebook 1's reasoning puzzle, Ultra thought about 4× longer than Nano and each answer cost about 50× more. Price your real workload, not a toy prompt.

### A.1 A latency wrapper for Chat Completions


Chat Completions doesn't report TTFT/t/s, but with streaming and a stopwatch, you can measure them yourself. This is genuinely how production clients do it:

````python
import time

def measure(model: str, prompt: str):
    """Measure TTFT, t/s, and E2EL for one streaming request."""
    t_start = time.perf_counter()
    t_first = None
    chunks = []

    stream = client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}],
        stream=True,
        stream_options={"include_usage": True},
    )

    usage = None
    for chunk in stream:
        if chunk.usage:
            usage = chunk.usage
            continue
        delta = chunk.choices[0].delta.content
        if delta:
            if t_first is None:
                t_first = time.perf_counter()   # <-- the moment the first token lands
            chunks.append(delta)

    t_end = time.perf_counter()

    ttft = t_first - t_start
    e2el = t_end - t_start
    out_tokens = usage.completion_tokens if usage else len(chunks)
    tps = out_tokens / (t_end - t_first)

    print(f"model:  {model}")
    print(f"TTFT:   {ttft:.2f}s")
    print(f"t/s:    {tps:.1f}")
    print(f"E2EL:   {e2el:.2f}s   ({out_tokens} output tokens)")
    return "".join(chunks)

_ = measure(
    "nvidia/NVIDIA-Nemotron-3-Nano-30B-A3B",
    "Explain what an AI agent is in one paragraph.",
)
````

**Why `measure()` crashed: two catches on Token Factory.**

1. **Usage arrives on every chunk.** `stream_options={"include_usage": True}` is meant to add one final usage-only chunk, but Token Factory attaches `usage` to *every* streamed chunk (42 of 42 when I checked). `measure()` hits `continue` whenever `chunk.usage` is set, so it skips every chunk, never records a content token, and `t_first` stays `None`, hence the `TypeError`.
2. **Reasoning streams separately.** Nano sends its thinking in a `reasoning` field before any `content`. Even with catch 1 fixed, watching only `delta.content` would report *time to first answer token*, after all the thinking. And `completion_tokens` (which includes the reasoning) divided by only the answer's streaming time would inflate t/s.

Here's a version that handles both:

````python
def measure_v2(model: str, prompt: str):
    """Like measure(), but counts streamed reasoning too, so reasoning models report honest numbers."""
    t_start = time.perf_counter()
    t_first_any = t_first_answer = None
    chunks, usage = [], None

    stream = client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}],
        stream=True,
        stream_options={"include_usage": True},
    )
    for chunk in stream:
        if chunk.usage:
            usage = chunk.usage
        if not chunk.choices:
            continue
        delta = chunk.choices[0].delta
        thinking = (delta.model_extra or {}).get("reasoning")
        now = time.perf_counter()
        if (thinking or delta.content) and t_first_any is None:
            t_first_any = now                 # first token of any kind: the model is alive
        if delta.content:
            if t_first_answer is None:
                t_first_answer = now          # first token of the actual answer
            chunks.append(delta.content)

    t_end = time.perf_counter()
    out_tokens = usage.completion_tokens
    print(f"model:               {model}")
    print(f"TTFT (any token):    {t_first_any - t_start:.2f}s")
    print(f"TTFT (answer token): {t_first_answer - t_start:.2f}s")
    print(f"t/s:                 {out_tokens / (t_end - t_first_any):.1f}   (all {out_tokens} output tokens, reasoning included)")
    print(f"E2EL:                {t_end - t_start:.2f}s")
    return "".join(chunks)

_ = measure_v2(
    "nvidia/NVIDIA-Nemotron-3-Nano-30B-A3B",
    "Explain what an AI agent is in one paragraph.",
)
````

### A.2 A cost wrapper for every request

The per-request calculation from Section 2, packaged so every call reports its own price:

````python
def tracked_completion(**kwargs):
    """Run a chat completion and print what it cost."""
    resp = client.chat.completions.create(**kwargs)
    u = resp.usage
    cost = (u.prompt_tokens * PRICE_IN + u.completion_tokens * PRICE_OUT) / 1e6
    print(f"input tokens:  {u.prompt_tokens}")
    print(f"output tokens: {u.completion_tokens}")
    print(f"cost:          ${cost:.6f}")
    return resp, cost

resp, cost = tracked_completion(
    model="nvidia/NVIDIA-Nemotron-3-Nano-30B-A3B",
    messages=[{"role": "user", "content": "Summarize the outlook for the specialty coffee market in one paragraph."}],
)
````
